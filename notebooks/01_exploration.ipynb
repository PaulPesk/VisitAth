```python
import json, re
import numpy as np

# 1. Μπάνια από το κείμενο
def parse_baths(text):
    if pd.isna(text):
        return np.nan
    t = str(text).lower()
    m = re.search(r"(\d+(\.\d+)?)", t)
    if m:
        return float(m.group(1))
    return 0.5 if "half" in t else np.nan

df["baths"] = df["bathrooms_text"].apply(parse_baths)
df["shared_bath"] = df["bathrooms_text"].str.contains("shared", case=False, na=False).astype(int)

# 2. Superhost και κριτικές
df["is_superhost"] = df["host_is_superhost"].map({"t": 1, "f": 0})
df["has_reviews"] = (df["number_of_reviews"] > 0).astype(int)

# 3. Ανέσεις
def parse_amenities(s):
    try:
        return [a.lower() for a in json.loads(s)]
    except Exception:
        return []

df["amenity_list"] = df["amenities"].apply(parse_amenities)
df["n_amenities"] = df["amenity_list"].apply(len)

flags = {
    "has_wifi": "wifi",
    "has_ac": "air conditioning",
    "has_kitchen": "kitchen",
    "has_balcony": "balcony",
    "has_elevator": "elevator",
    "has_parking": "parking",
}
for col, kw in flags.items():
    df[col] = df["amenity_list"].apply(lambda L: int(any(kw in a for a in L)))

# 4. Απόσταση από το Σύνταγμα (km)
def haversine_km(lat1, lon1, lat2, lon2):
    R = 6371
    p1, p2 = np.radians(lat1), np.radians(lat2)
    dphi = p2 - p1
    dl = np.radians(lon2 - lon1)
    a = np.sin(dphi / 2) ** 2 + np.cos(p1) * np.cos(p2) * np.sin(dl / 2) ** 2
    return 2 * R * np.arcsin(np.sqrt(a))

df["dist_center_km"] = haversine_km(df["latitude"], df["longitude"], 37.9755, 23.7348)

df[["baths", "is_superhost", "n_amenities", "has_ac", "dist_center_km"]].describe()
```
